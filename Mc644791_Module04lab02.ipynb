```python
# TASK1：数据准备
# 导入所需包
import pandas as pd
import plotly.express as px
import dash
from dash import dcc
from dash import html

# 将航空公司数据读取到 pandas 数据框中
airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str})

# 随机抽取 500 个数据点。将随机数状态设置为 42，以便获得相同的结果。
data = airline_data.sample(n=500, random_state=42)

# 创建饼图
fig = px.pie(data, values='Flights', names='DistanceGroup', title='按航班划分的距离组比例')

# TASK2：创建Dash应用
app = dash.Dash(__name__)

# TASK3 + TASK4 + TASK5：布局（标题、描述、饼图）
app.layout = html.Div(children=[
    # TASK3 标题 Airline Dashboard，样式要求
    html.H1('Airline Dashboard',
            style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    # TASK4 应用描述文本
    html.P('Proportion of distance group (250 mile distance interval group) by flights.',
           style={'textAlign':'center', 'color':'#F57241'}),
    # TASK5 添加图表
    dcc.Graph(figure=fig)
])

# TASK6 运行程序
if __name__ == '__main__':
    app.run(port=8051)

```
